```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# Load clustered dataset
df = pd.read_csv(r"C:\Users\Sam\Desktop\Spotify_Genre_Segmentation\data\output\spotify_clustered.csv")
```

```python

# -------------------------------
# 1. Cluster Distribution
# -------------------------------
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="cluster")
plt.title("Cluster Distribution (K=5)")
plt.show()
```

```python

# -------------------------------
# 2. Cluster vs Playlist Genre
# -------------------------------
plt.figure(figsize=(12,6))
sns.countplot(data=df, x="playlist_genre", hue="cluster")
plt.xticks(rotation=45)
plt.title("Cluster Distribution across Playlist Genres")
plt.show()
```

```python
# Top 15 playlist names
top_playlists = df["playlist_name"].value_counts().head(15).index
df_top = df[df["playlist_name"].isin(top_playlists)]

plt.figure(figsize=(14,7))
sns.countplot(data=df_top, y="playlist_name", hue="cluster")
plt.title("Top 15 Playlist Names vs Clusters")
plt.show()
```
