```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# Load cleaned dataset
df = pd.read_csv(r"C:\Users\Sam\Desktop\Spotify_Genre_Segmentation\data\processed\spotify_cleaned.csv")
```

```python

print("Shape:", df.shape)
```

```python
print(df.head())
```

```python

# -------------------------------
# 1. Genre Distribution (Countplot)
# -------------------------------
plt.figure(figsize=(12,6))
sns.countplot(data=df, x="playlist_genre")
plt.xticks(rotation=45)
plt.title("Genre Distribution")
plt.show()
```

```python

# -------------------------------
# 2. Histogram of Numeric Features
# -------------------------------
df.hist(figsize=(18,12), bins=20)
plt.suptitle("Histogram of Numeric Features")
plt.show()
```

```python

# -------------------------------
# 3. Boxplot (Outlier detection)
# -------------------------------
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x=df["danceability"])
plt.title("Boxplot of Danceability")
plt.show()
```

```python

# -------------------------------
# 4. Scatter Plot (Energy vs Danceability)
# -------------------------------
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x="danceability", y="energy", hue="playlist_genre")
plt.title("Danceability vs Energy by Genre")
plt.show()
```

```python

# -------------------------------
# 5. Pairplot (Important features)
# -------------------------------
important_features = ["danceability", "energy", "loudness", "tempo", "valence", "playlist_genre"]
sns.pairplot(df[important_features], hue="playlist_genre")
plt.show()
```
