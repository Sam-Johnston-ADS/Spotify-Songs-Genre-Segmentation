```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\Sam\Desktop\Spotify_Genre_Segmentation\data\raw\spotify dataset.csv")
```

```python
"Before Cleaning Shape:", df.shape
```

```python
"After Removing Duplicates Shape:", df.shape
```

```python
"\nMissing values before cleaning:\n", df.isnull().sum()
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
categorical_cols = df.select_dtypes(include=["object"]).columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
"\nMissing values after cleaning:\n", df.isnull().sum()
```

```python
df.to_csv(r"C:\Users\Sam\Desktop\Spotify_Genre_Segmentation\data\processed\spotify_cleaned.csv", index=False)

print("✅ Cleaned dataset saved successfully!")
```
